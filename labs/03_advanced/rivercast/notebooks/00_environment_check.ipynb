# 00 — Environment check

Verifies that this workbench (or any development machine) is ready for the
RiverCast lab: package import, configuration, writable storage, fixtures,
and optional services (MLflow, KFP, cluster API).

**How to run:** select the *Python (rivercast)* kernel, then
*Restart Kernel and Run All Cells*. All logic lives in
`rivercast.envcheck` — this notebook only calls it and shows the results.

> RiverCast is an **educational** system, not a flood-warning product.

**Reading the results**

| Status | Meaning |
|---|---|
| `PASS` | verified working |
| `WARN` | not available/configured yet — acceptable at this phase (fixtures arrive in Phase 2; MLflow/KFP/cluster checks warn off-cluster) |
| `FAIL` | broken — fix before continuing; the last cell raises on any `FAIL` |

```python
from pathlib import Path

from rivercast import envcheck
from rivercast.log import configure_logging

configure_logging()
LAB_ROOT = envcheck.find_lab_root(Path.cwd())
CONFIG_PATH = LAB_ROOT / "configs" / "local.yaml"
print(f"lab root : {LAB_ROOT}")
print(f"config   : {CONFIG_PATH.relative_to(LAB_ROOT)}")
```

```python
results = envcheck.run_all(config_path=CONFIG_PATH, lab_root=LAB_ROOT)
print(envcheck.summarize(results))
```

```python
# Raises RuntimeError if anything above is FAIL — a clean run-all means the
# environment is ready for the next notebook.
envcheck.require_no_failures(results)
print("Environment ready.")
```
